## 2. 數字和字串不一樣

有引號的是 **字串**，電腦把它當成「一串字」；沒有引號的是 **數字**，電腦會幫你計算。

**先猜猜看**：下面三行會分別印出什麼？想好再執行。

````python
print(3 + 5)      # 沒有引號：這是數字，電腦會算出答案
print("3" + "5")  # 有引號：這是兩個字串，+ 會把它們「接起來」
print("3 + 5")    # 整個都在引號裡：原封不動顯示出來
````

你猜對了嗎？

- `3 + 5` → `8`（數字相加）
- `"3" + "5"` → `35`（字串接在一起）
- `"3 + 5"` → `3 + 5`（照字面顯示）

> **注意**：程式裡的引號、括號、逗號都要用 **英文半形**。打成中文的「」或（）會出錯。

## 3. 變數：幫資料取名字

在 Scratch 裡，你會建立一個「分數」變數來記分。Python 的寫法更簡單：

```
變數名稱 = 要存的值
```

這裡的 `=` 不是數學的「等於」，而是 **「把右邊的值，存進左邊的名字」**，就像 Scratch 的「將 ___ 設為 ___」。

````python
name = "小安"   # 把字串 "小安" 存進變數 name
age = 13        # 把數字 13 存進變數 age

# print 可以一次印好幾樣東西，用逗號隔開，中間會自動空一格
print("我叫", name)
print("我今年", age, "歲")
````

變數的值可以改變。下面就是 Scratch 的「將 分數 改變 10」：

````python
score = 0              # 遊戲開始，分數設為 0
print("開始：", score)

score = score + 10     # 先算右邊：目前的 score（0）加 10，再存回 score
print("吃到金幣：", score)

score = score + 10     # 再加一次 10
print("又吃到金幣：", score)
````

**取變數名字的規則**
- 可以用英文字母、數字、底線 `_`，但 **不能用數字開頭**（`2name` 不行，`name2` 可以）
- 大小寫不一樣：`Score` 和 `score` 是兩個不同的變數
- 取個看得懂的名字：`price` 比 `p` 好懂

## 4. 讓 Python 幫你算數學

| 符號 | 意思 | 例子 | 結果 |
| --- | --- | --- | --- |
| `+` | 加 | `7 + 2` | `9` |
| `-` | 減 | `7 - 2` | `5` |
| `*` | 乘（不是 ×） | `7 * 2` | `14` |
| `/` | 除 | `7 / 2` | `3.5` |
| `//` | 除完只取整數（商） | `7 // 2` | `3` |
| `%` | 除完的餘數 | `7 % 2` | `1` |

`//` 和 `%` 是新朋友，考 APCS 時很常出現。用分糖果來想：**7 顆糖分給 2 個人，每人 3 顆（`7 // 2`），剩 1 顆（`7 % 2`）。**

````python
candy = 7     # 有 7 顆糖
people = 2    # 分給 2 個人

print("每人分到：", candy // people, "顆")   # 商：每人幾顆
print("還剩下：", candy % people, "顆")      # 餘數：剩幾顆
print("如果可以切開，每人：", candy / people, "顆")
````
