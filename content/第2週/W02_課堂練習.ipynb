在 Scratch 裡，你用過「如果…那麼…否則」積木。Python 也一樣，只是把積木換成文字：

| Scratch 積木 | Python 寫法 |

| 否則 | `else:` |
| 積木「包住」裡面的積木 | 往右 **縮排 4 格** |
| <（ ）=（ ）> | `==`（兩個等號） |

上週學的 `print`、變數和 `=` 都會用到。記得先執行下面的小幫手。

````python
# ▶ 請先執行這一格（點一下這格，再按 Shift + Enter）
# 這格會載入「檢查小幫手」，後面的練習可以用它看看自己有沒有做對。
# 你不用看懂這格的內容，第 5 週學到「函式」時就會懂了。

def check(ok, hint="再看一次題目，改一改再執行！"):
    """ok 是 True 就說答對，是 False 就給提示。"""
    if ok:
        print("○ 答對了！")
    else:
        print("× 還沒對喔。" + hint)

print("小幫手載入完成，可以開始了！")
````

## 1. 是或不是：True 和 False

電腦判斷事情只有兩種答案：**True（成立）** 或 **False（不成立）**。
就像 Scratch 裡六角形的條件積木。

| 符號 | 意思 |
| --- | --- |
| `>` `<` | 大於、小於 |
| `>=` `<=` | 大於等於、小於等於 |
| `==` | 等於（**兩個**等號） |
| `!=` | 不等於 |

**先猜猜看**再執行：

````python
print(5 > 3)            # 5 大於 3 嗎？
print(10 == 10)         # 10 等於 10 嗎？
print(7 != 7)           # 7 不等於 7 嗎？
print("布" == "布")     # 字串也可以比較：兩個字一樣嗎？
print("布" == "石頭")
````

> **最常見的錯誤**：`=` 和 `==` 搞混。
> - `score = 10`：把 10 **存進** score
> - `score == 10`：**問** score 是不是 10，答案是 True 或 False

## 2. if：如果…那麼

```
if 條件:
    條件成立時要做的事
```

注意兩件事：
1. 條件後面一定要有 **冒號** `:`
2. 要做的事要往右 **縮排 4 格**（按一次 Tab 鍵）。縮排就是 Scratch 積木「被包在裡面」的意思。

````python
temperature = 33     # 今天的氣溫

if temperature > 30:
    print("好熱！記得多喝水。")   # 有縮排：只有條件成立才會執行

print("今天氣溫", temperature, "度")  # 沒縮排：不管怎樣都會執行
````

把 `temperature` 改成 `25` 再執行一次，看看哪一行不見了。

## 3. if / else：如果…那麼…否則

````python
score = 58

if score >= 60:
    print("及格了！")
else:
    print("還差一點，下次加油！")   # 條件不成立時，就執行 else 裡面的
````

## 4. elif：還有其他可能

答案不只兩種時，用 `elif`（else if 的縮寫），意思是「不然的話，如果…」。
Python 會 **由上往下** 檢查，**第一個成立的就執行，後面的都跳過**。

````python
drink = "奶茶"

if drink == "紅茶":
    print("紅茶 25 元")
elif drink == "奶茶":
    print("奶茶 35 元")
elif drink == "綠茶":
    print("綠茶 25 元")
else:
    print("沒有這個飲料喔")
````

## 5. and、or：組合條件

- `and`（而且）：兩邊 **都** 成立才是 True
- `or`（或者）：**只要一邊** 成立就是 True

````python
is_weekend = True
is_sunny = False

if is_weekend and is_sunny:
    print("去公園打球！")
else:
    print("在家寫程式。")

day = "六"
if day == "六" or day == "日":
    print("今天是週末")
````

## 6. 【實作】猜拳裁判

規則你一定知道：**剪刀贏布、石頭贏剪刀、布贏石頭**，出一樣就平手。

我們先把判斷想清楚，再寫程式：
1. 如果兩個人出的一樣 → 平手
2. 不然，如果是「玩家贏」的三種情況之一 → 玩家贏
3. 其他情況 → 電腦贏

第 2 步的三種情況，可以用 `or` 連起來，每一種情況裡面用 `and`。
請完成下面的程式，讓 `result` 存放正確的結果。

````python
player = "石頭"      # 玩家出什麼（可以改成 "剪刀"、"石頭"、"布"）
computer = "剪刀"    # 電腦出什麼

if player == computer:
    result = "平手"
# 在下面補上：玩家贏的三種情況（用 or 連接）
elif (player == "剪刀" and computer == "布"):
    result = "玩家贏"
else:
    result = "電腦贏"

print("玩家出", player, "，電腦出", computer, "→", result)
check(result == "玩家贏", "提示：石頭贏剪刀這種情況還沒寫進去喔")
````

## 7. 讓電腦自己出拳

真正的對戰，電腦不能先讓你知道它出什麼。`random` 是 Python 內建的「骰子工具」，`random.choice(...)` 會從一串東西裡 **隨機挑一個**。

中括號 `[ ]` 叫做 **清單**，第 4 週會仔細學，今天先照著用就好。

每執行一次，電腦出的拳都可能不一樣，多按幾次 Shift + Enter 試試看。

````python
import random     # 載入隨機工具（像 Scratch 的「隨機取數」積木）

player = "布"    # 改成你想出的拳
computer = random.choice(["剪刀", "石頭", "布"])   # 電腦隨機出拳

if player == computer:
    result = "平手"
elif (player == "剪刀" and computer == "布") or \
     (player == "石頭" and computer == "剪刀") or \
     (player == "布" and computer == "石頭"):
    result = "你贏了！"
else:
    result = "電腦贏了！"

print("你出", player, "，電腦出", computer, "→", result)
````

## 8. 【挑戰關】防呆裁判

如果有人亂出拳，例如 `player = "火箭"`，現在的程式會說「電腦贏了」，這不公平。
請在最前面加一個判斷：如果 `player` **不是** 剪刀、石頭、布其中之一，`result` 就設為 `"出拳無效"`。

````python
player = "火箭"
computer = "布"

# 在最前面加一個 if，判斷 player 是不是合法的拳，下面原本的 if 改成 elif
if player == computer:
    result = "平手"
elif (player == "剪刀" and computer == "布") or \
     (player == "石頭" and computer == "剪刀") or \
     (player == "布" and computer == "石頭"):
    result = "玩家贏"
else:
    result = "電腦贏"

print(result)
check(result == "出拳無效", "提示：if player != \"剪刀\" and player != \"石頭\" and player != \"布\":")
````

## 今天學到的

- `if` / `elif` / `else` 後面要有冒號 `:`，裡面要縮排 4 格
- `==` 是「問是不是一樣」，`=` 是「存進去」
- `and` 要兩邊都成立，`or` 只要一邊成立
